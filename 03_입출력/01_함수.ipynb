```python
# 파이썬 함수의 구조
# def 함수이름(매개변수1, 매개변수2, ...):
# 수행할 문장1
# 수행할문장2
# ...

def add(a, b):
	return a + b
```

```python
# (a: Any, b: Any) -> any
result = add(10, 20)
print(result)
```

```python
# 1. 일반적인 함수
# 입력값이 있고, 반환값이 있다.

def add(a, b):
	result = a + b
	return result
sum = add(3, 4)
print(sum)
```

```python
# 2. 입력값이 없는 함수
# 입력값 X, 반환값 O

def say():
	return "hi"

a = say()
print(a)
```

```python
# 3. 반환값이 없는 함수
# 입력값 O, 반환값 X return이 없기 때문. sum에 아무것도 안감.
def add(a, b):
	print(f"{a}, {b}의 합은 {a+b}입니다.")

sum = add(3, 4)
print(sum)
```

```python
# 4. 입력값도 없고, 반환값도 없는 함수
def say():
	print("hi")

a = say()
print(a)
```

```python
# 매개변수를 지정하여 호출하기
def sub(a, b):
	return a - b

result = sub(a=7, b=3)
print(result)

result2 = sub(b=7, a=3)
print(result2)

print("안녕", "바이", sep="---")
```

```python
# 입력값이 몇 개가 될지 모를때는 어떻게 할까?
def add_many(*args):
	result = 0
	print(type(args))
	for i in args:
		result = result + i
	return result

result = add_many(1, 2, 3, 4, 5)
print(result)
print()
```
