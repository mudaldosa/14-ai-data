```python
### 문제 4-1: 영화관 입장 조건

# 나이와 보호자 동반 여부를 입력받아 영화 관람 가능 여부를 판단하세요.

# - 12세 이상이면 관람 가능
# - 12세 미만이지만 보호자가 동반하면 관람 가능
# - 그 외는 관람 불가
  

age = int(input("나이를 입력하세요: "))
guardian = input("보호자 동반 여부 (yes/no): ")

# 힌트: guardian == "yes"를 활용하세요

if age >= 12 : print("관람가능")
elif  guardian == "yes" : print("관람가능")
else : print("관람불가")
```

```python
# 거리(km)와 시간대를 입력받아 요금을 계산하세요.

# - 기본요금: 3000원 (2km까지)
# - 추가요금: 2km 초과시 1km당 1000원
# - 심야할증(22시~04시): 총 요금의 20% 추가
distance = float(input("거리를 입력하세요 (km): "))
hour = int(input("현재 시간을 입력하세요 (0-23): "))
#   **예상 출력:**
#   거리를 입력하세요 (km): 5
#   현재 시간을 입력하세요 (0-23): 23
#   요금: 7200원

if distance <= 2 : avg = 3000
else : avg = 3000 + (distance-2)*1000

if hour >= 22 or hour <= 4 : avg = avg * 1.2
print(f"요금: {int(avg)}원")
```

```python
todos = []
priorities = []

while True:
	print("\n [나의 할 일 관리자]")
	print("1.추가 / 2.보기 / 3.수정 / 4.삭제 / 5.종료")
	choice = input("선택: ")
	
	if choice == "1":
		todo = input("할 일을 입력하세요: ")
		p = input("중요한가요? (별표 표시하려면 1 입력): ")
		todos.append(todo)
		if p == "1":
			priorities.append("#")
		else: 
			priorities.append(" ")
		print("등록 완료!")

	elif choice == "2":
		if len(todos) == 0:
			print("등록된 할 일이 없습니다.")
		else:
			for i in range(len(todos)):
				print(f"{i+1}.{priorities[i]}{todos[i]}")
	
	elif choice == "3":
		idx = int(input("수정할 번호:"))-1
		if 0 <= idx < len(todos):
			new_todo = input("새 할 일을 입력하세요: ")
			todos[idx] = new_todo
			print("수정 완료!")
		else:
			print("잘못된 번호입니다.")
	elif choice == "4":
		idx = int(input("삭제할 번호:"))-1
		if 0 <= idx < len(todos):
			todos.pop(idx)
			priorities.pop(idx)
			print("삭제 완료!")
		else:
			print("잘못된 번호입니다.")
	elif choice == "5":
		print("프로그램을 종료합니다.")
		break

```

```python
odd = []
even = []

for i in range(1, 21):
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print(f"홀수: {', '.join(map(str, odd))} (총 {len(odd)}개)")
print(f"짝수: {', '.join(map(str, even))} (총 {len(even)}개)")
```

```python
height = int(input("높이를 입력하세요: "))

if height < 1 or height > 10:
    print("1~10 사이만 입력 가능합니다.")
else:
    for i in range(1, height + 1):
        print(" " * (height - i), end="")  # end=""는 줄바꿈 방지임

        for j in range(1, i + 1):
            print(j, end="")

        for j in range(i - 1, 0, -1):
            print(j, end="")

        print()
```

```python
n = int(input("숫자를 입력하세요: "))

if n < 2:
    print("2 이상의 숫자를 입력하세요.")
else:
    is_prime = True
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print(f"{n}은(는) 소수입니다.")
    else:
        print(f"{n}은(는) 소수가 아닙니다.")

    primes = []
    for num in range(2, n + 1):
        prime = True
        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
        if prime:
            primes.append(num)

    print("\n1부터", n, "까지의 소수:")
    print(", ".join(map(str, primes)))
    print(f"총 {len(primes)}개의 소수가 있습니다.")
```

```python
balance = 10000

while True:
    print("\n=== ATM 시스템 ===")
    print("1. 잔액 조회")
    print("2. 입금")
    print("3. 출금")
    print("4. 종료")

    choice = int(input("선택: "))

    if choice == 1:
        print(f"현재 잔액: {balance}원")

    elif choice == 2:
        money = int(input("입금액: "))
        balance += money
        print(f"입금 완료! 현재 잔액: {balance}원")

    elif choice == 3:
        money = int(input("출금액: "))
        if money > balance:
            print(f"잔액이 부족합니다. 현재 잔액: {balance}원")
        else:
            balance -= money
            print(f"출금 완료! 현재 잔액: {balance}원")

    elif choice == 4:
        print("이용해 주셔서 감사합니다.")
        break

    else:
        print("잘못된 선택입니다.")
```
