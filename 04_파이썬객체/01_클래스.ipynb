```python
class Calculator:
	def __init__(self):
		self.result = 0

	def add(self, num):
		self.result = self.result + num
		return self.result
	
	def sub(self, num):
		self.result -= num
		return self.result
	
	def mul(self, num):
		self.result *= num
		return self.result

	def dev(self, num):
		self.result /= num
		return self.result
```

```python
# cal1 10 더하고 2 뺴고 5 곱하고 4나눈값  #cal2 5 더하고 10 곱하고 10 뺴고 5 나눈값
cal1 = Calculator()
print(cal1.add(10))
print(cal1.sub(2))
print(cal1.mul(5))
print(cal1.dev(4))

cal2 = Calculator()
print(cal2.add(5))
print(cal2.sub(10))
print(cal2.mul(10))
print(cal2.dev(5))
```
