```python
#딕셔너리 관련된 함수

#1. key 리스트 만들기 = keys
profile = {"이름":"김말똥","나이":"57","번호":"12345"}
print(profile.keys())

#2. value 리스트 만들기 = values
profile = {"이름":"김말똥","나이":"57","번호":"12345"}
print(profile.values())

#3. key와value 쌍으로 넣기
print(profile.items())

#4. key,value 쌍 모두 지우기 = clear
profile = {"이름":"김말똥","나이":"57","번호":"12345"}
print(profile.clear())

#5. key로 value 얻기 = get
profile = {"이름":"김말똥","나이":"57","번호":"12345"}

#6. 해당 key가 딕셔너리 안에 있는지 조사하기 = in
profile = {"이름":"김말똥","나이":"57","번호":"12345"}
print("이름" in profile)

#7. key로 value 얻기 = pop
profile = {"이름":"김말똥","나이":"57","번호":"12345"}
print(profile)
print(profile.pop("이름"))

```
