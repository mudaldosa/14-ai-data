### 문자열 포맷 코드

```python

# 3. 2개 이상의 값 넣기
number = 10
day = "three"
print("I ate {0} apples, so I was sick for {1} days".format(number, day))
```

```python
# f문자열 포매팅
name = "홍길동"
age = 30
print(f"나의 이름은 {name}입니다. 나이는 {age}입니다")
```

```python
# 문자열 관련 함수들

# 1. 문자 개수 세기 . count

a = "hobby"
print(a.count("b"))

# 2. 위치 알려주기 - find
print(a.find("b"))
print(a.find("k"))

# 3. 위치 알려 주기  . index
print(a.index("b"))

# 4. 문자열 삽입- join
print(",".join("abcd"))

# 5. 소문자를 대문자로 바꾸는 함수 - upper
a = "hello"
print(a.upper())

#6. 대문자를 소문자로 바꾸는 함수 - lower
a = "HI"
print(a.lower())

# 7. 공백지우기 strip
a = "  hi   "
print(a)
print(a.strip())

# 8. 왼쪽 공백지우기 lstrip
# 9. 오른쪽 공백지우기 rstrip
print("왼쪽", a.lstrip())
print("오른쪽", a.rstrip())
```
