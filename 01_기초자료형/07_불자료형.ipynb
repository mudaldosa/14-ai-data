```python
# 1. 다음 정보를 담는 딕셔너리를 만드시오
#    - 이름: 본인 이름
#    - 나이: 본인 나이
#    - 취미: 본인 취미 (리스트로 여러 개)
#    - 거주지: 본인 거주지

# 2. 위 딕셔너리에 '전공' 항목을 추가하시오

# 3. '취미' 항목에 새로운 취미를 추가하시오

# 4. '나이' 항목의 값을 1 증가시키시오

# 5. 모든 key와 value를 출력하시오

# 1
me = {
    '이름': '백무성',
    '나이': 26,
    '취미': ['게임', '영화'],
    '거주지': '대구'
}

# 2
me['전공'] = '기계공학'

# 3
me['취미'].append('운동')

# 4
me['나이'] += 1

# 5
for j, k in me.items():
    print(j)
```

```python
# 다음 학생들의 성적 정보를 딕셔너리로 저장하시오
# 학생1: 이름=김민수, 국어=85, 영어=90, 수학=78
# 학생2: 이름=이영희, 국어=92, 영어=88, 수학=95
# 학생3: 이름=박철수, 국어=78, 영어=85, 수학=82

# 요구사항:
# 1. 각 학생의 총점과 평균을 계산하여 딕셔너리에 추가
# 2. 각 학생의 정보를 보기 좋게 출력

students = [
    {'이름': '김민수', '국어': 85, '영어': 90, '수학': 78},
    {'이름': '이영희', '국어': 92, '영어': 88, '수학': 95},
    {'이름': '박철수', '국어': 78, '영어': 85, '수학': 82}
]

print("=== 학생 성적 관리 ===")

for s in students:
    total = s['국어'] + s['영어'] + s['수학']
    avg = total / 3
    
    s['총점'] = total
    s['평균'] = avg

    print(f"이름: {s['이름']}")
    print(f"국어: {s['국어']}점, 영어: {s['영어']}점, 수학: {s['수학']}점")
    print(f"총점: {total}점, 평균: {avg:.3f}점")
    print("-----------------------")
print(len(students))
```
